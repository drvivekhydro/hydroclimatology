## Classes and objects

A **class** bundles data (*attributes*) together with the functions that act on that
data (*methods*). You have already used classes without noticing: a NumPy array, a
pandas `DataFrame`, and an xarray `Dataset` are all objects with attributes (`.shape`,
`.columns`) and methods (`.mean()`, `.plot()`).

Writing your own class is useful whenever a thing has both *state* and *behaviour*.
A weather station, for example, has a location and a growing record of daily values,
and we want to ask it for its mean or its number of hot days.


````python
class Station:
    \"\"\"A weather station and its record of daily mean temperature.\"\"\"

    def __init__(self, name, lat, lon):
        self.name = name          # attribute
        self.lat = lat
        self.lon = lon
        self.temperature = []     # daily values [degC], filled in later

    def add_day(self, value):     # method
        self.temperature.append(value)

    def mean(self):
        return sum(self.temperature) / len(self.temperature)

    def hot_days(self, threshold=30.0):
        return sum(1 for t in self.temperature if t > threshold)


mandi = Station("Mandi", 31.71, 76.93)
for t in [28.4, 31.1, 33.0, 29.7, 30.6]:
    mandi.add_day(t)

print(mandi.name, "mean:", round(mandi.mean(), 2), "degC")
print("hot days (>30):", mandi.hot_days())
````

The `__init__` method runs when the object is created and sets up its attributes.
`self` is the object itself, passed automatically, so `mandi.add_day(31.1)` becomes
`Station.add_day(mandi, 31.1)`.

This is exactly the pattern of the `EBM` class you meet in **Lecture 4**: its state is
the temperature and the current year, and its method steps that state forward one year
at a time. Recognising the pattern here makes that notebook easier to read.

## 1D Plot

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

x = np.arange(0,300) # Numbers from 0 to 30 (30 not included)
y = np.sin(np.arange(0,300)) + np.random.normal(0,5,300) # Create sinus
                                                       ##signal and add noise

dataFrame = pd.DataFrame({'Intervals': x,
                         'values': y})    

rolling_mean = dataFrame['values'].rolling(10, min_periods=1, center=True).mean()

plt.plot(x,y,label='Kurve', alpha=0.5)
plt.plot(x, rolling_mean.values, label = 'Rolling Mean')
plt.title('Sinus + Random')
plt.xlabel('Interval')
plt.ylabel('Values')
plt.legend()
````


## Gridded data: NetCDF and xarray

Climate and reanalysis data come as multi-dimensional arrays — a variable on a
latitude–longitude–time grid — stored in **NetCDF** files. **xarray** opens these and
lets you select, average, and plot by *named* dimension instead of by array index.

We use a small built-in example dataset: 2-metre air temperature from the NCEP/NCAR
reanalysis over North America (four times daily, 2013–2014). Everything below works
the same way on an ERA5, IMD, or CMIP6 file.


````python

import xarray as xr

ds = xr.tutorial.open_dataset("air_temperature")   # downloads ~4 MB on first use
ds
````

````python

# 'air' is 2-m air temperature in kelvin; convert to degC and inspect one time step
air_c = ds["air"] - 273.15
air_c.isel(time=0)
````

````python

# a map: the first time step
air_c.isel(time=0).plot(cmap="coolwarm")
````

````python

# a regional-mean time series: average over a lon-lat box
region = air_c.sel(lat=slice(50, 30), lon=slice(240, 260))   # roughly the US Great Plains
region_mean = region.mean(dim=["lat", "lon"])
````

````python
region_mean.plot()
````

````python

# a smoothed version
region_mean.rolling(time=20, min_periods=1).mean().plot()
````
